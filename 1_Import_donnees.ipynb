VISUALISATION EXCEL FILE 

```python

import os
import h5py
import numpy as np
import pandas as pd
import scipy.io as sio

# Définition des chemins d'accès
DOSSIER_PROJET = r"C:\Users\ReMarkt\Documents\0_PROJET_SCIENTIST\Data_CWRU\Sains"
FICHIER_MAT = os.path.join(DOSSIER_PROJET, "100.mat")
FICHIER_EXCEL = os.path.join(DOSSIER_PROJET, "100_mat.xlsx")


def mat_to_excel(mat_file_path, excel_file_path):
  print(f"Lecture du fichier : {mat_file_path}")

  # 1. Lecture du fichier .mat selon sa version
  try:
    data = sio.loadmat(mat_file_path)
    variables = {k: v for k, v in data.items() if not k.startswith("__")}
    print("Format standard détecté (SciPy).")
  except Exception:
    with h5py.File(mat_file_path, "r") as f:
      variables = {k: np.array(v) for k, v in f.items()}
    print("Format v7.3 détecté (H5Py).")

  # 2. Écriture dans le fichier Excel
  print(f"Création du fichier Excel : {excel_file_path}")
  with pd.ExcelWriter(excel_file_path, engine="openpyxl") as writer:
    for var_name, matrix in variables.items():
      # Gestion des dimensions pour créer un DataFrame propre
      if matrix.ndim == 1 or (matrix.ndim == 2 and matrix.shape[1] == 1):
        df = pd.DataFrame(matrix, columns=[var_name])
      elif matrix.ndim == 2:
        df = pd.DataFrame(matrix)
      else:
        df = pd.DataFrame(matrix.reshape(matrix.shape[0], -1))

      # Limitation à 31 caractères max pour le nom de l'onglet Excel
      sheet_name = var_name[:31]
      df.to_excel(writer, sheet_name=sheet_name, index=False)

  print("Conversion terminée avec succès !")


# Lancement du script
if os.path.exists(FICHIER_MAT):
  mat_to_excel(FICHIER_MAT, FICHIER_EXCEL)
else:
  print(
      f"Erreur : Le fichier '98.mat' est introuvable dans le dossier :\n{DOSSIER_PROJET}"
  )
```
